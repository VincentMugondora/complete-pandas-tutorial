## Intro to Dataframes

```python
import pandas as pd

df = pd.DataFrame([[1,2,3],[4,5,6],[7,8,9]], columns=['A','B','C'], index=['a','b','c'])

print(df)
```

```python
df.head()
```

```python
print(df.tail())
print(df.describe())
print(df.info())
print(df.columns)
print(df.index)
print(df.dtypes)
print(df.shape)
print(df.values)
print(df.T)
print(df.to_numpy())
print(df.to_dict())
print(df.to_string(index=False))
print(df.to_csv(index=False))
# print(df.to_json(orient='records'))
# print(df.to_html(index=False))
# print(df.to_markdown(index=False))
# print(df.to_latex(index=False))
print(df.to_clipboard(index=False))
# print(df.to_excel('output.xlsx', index=False))
```
